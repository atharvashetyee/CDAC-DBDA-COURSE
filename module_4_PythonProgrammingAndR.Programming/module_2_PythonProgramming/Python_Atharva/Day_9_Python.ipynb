```python
import re
p=r'\w'
print("Only allowed is A to Z (upper and lower), 0 to 9 (digits), and underscore.")
user_name=input("Enter user name:")

if re.search(p,user_name):
    print(f"{user_name} is accepted")
else:
    print(f"{user_name} can not have digit")
```

```python
import re
p=r'^\d'
print("Only allowed is A to Z (upper and lower), 0 to 9 (digits), and underscore.")
user_name=input("Enter user name:")

if re.search(p,user_name):
    print(f"{user_name} is accepted")
else:
    print(f"{user_name} must have digit")
```

```python
import re
p=r'\d{10}'
print("Only allowed is A to Z (upper and lower), 0 to 9 (digits), and underscore.")
mobile_no=input("Enter Mobile Number:")

if re.search(p,mobile_no):
    print(f"{mobile_no} is accepted")
else:
    print(f"{mobile_no} must have 10 digit")
```
