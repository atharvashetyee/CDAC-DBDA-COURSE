```python
#7) Reverse a string
def reversed(list1:str):
  word=list1.split()
  for i in word:
    new=word[::-1]
  print(new)

```

```python
#8) Count vowels in a string
def count_vowel(word):
  word1=word.lower()

  sum=0
  count_vowels=["a","e","i","o","u"]
  for i in word1:

    if i in count_vowels:
      sum+=1
    else:
      sum+=0
  print(sum)
```

```python
#13) Remove duplicates from a list (preserve order)
def remove_dep(input):
  new=[]
  for i in input:
    if i not in new:
      new.append(i)
  return new
```

```python
remove_dep([1,22,22,3,4])
```
