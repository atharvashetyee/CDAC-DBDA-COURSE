```python
menu_page = '''
Item Code   Item Name         Price (₹)
1           Cutting_Chai      15.00
2           Masala_Chai       25.00
3           Samosa            20.00
4           Vada_Pav          30.00
5           Biscuit_Packet    10.00
'''
print(menu_page)
menu = {"cutting_chai":15, "masala_chai":25, "samosa":20, "vada_pav":30, "biscuit_packet":10}

print('''==== Please enter item name EXACTLY AS MENTIONED IN MENU and it's quantity in the following example format : "cutting_chai 5 samosa 2 biscuit 1". ====
==== Please note that there should not be any space in between the item name ==== ''')

new_order = input("enter your order ---> ").lower()

membership = int(input('''Enter 1 if you have Gold membership
Enter 2 if you have Silver membership
Enter 0 if you don't have any membership ------> '''))

order_list = new_order.split()

bill = {}

# calculating individual item price according to quantity
for i in order_list:
  if order_list.index(i) % 2 == 0:
    item_price = int(order_list[order_list.index(i) + 1]) * menu[i]
    bill[i] = item_price

subtotal = 0
for i in bill:
  subtotal += bill[i]

# deciding amount of discount
if membership == 1:
  memb_name = "Gold(10% off)"
  taxable_total = subtotal - (subtotal*0.1)
  discount_applied = "-" + str(subtotal*0.1)
elif membership == 2:
  memb_name = "Silver(5% off)"
  taxable_total = subtotal - (subtotal*0.05)
  discount_applied = "-" + str(subtotal*0.05)

# applying tax
gst = "+" + str(taxable_total*0.05)
total_amount_due = taxable_total + (taxable_total*0.05)

# printing final bill header
print('''
=========================================
               TOTAL BILL
=========================================''')

print("ITEM".ljust(20," "),"QTY    PRICE TOTAL")

print("-----------------------------------------   ")

# printing individual metrics
for i in order_list:
  if order_list.index(i) % 2 == 0:
    print(i.ljust(20," "),f"{int(order_list[order_list.index(i) + 1])}      {menu[i]}    {(int(order_list[order_list.index(i) + 1])) * (menu[i])}")

# printing total metrics

print(f'''
-----------------------------------------
SUBTOTAL                             {subtotal}
MEMBERSHIP: {memb_name}
DISCOUNT APPLIED                     {discount_applied}
-----------------------------------------
TAXABLE TOTAL                        {taxable_total}
GST (5.00%)                          {round((float(gst)),2)}
-----------------------------------------
TOTAL AMOUNT DUE                     {total_amount_due}
=========================================
THANK YOU! VISIT AGAIN!
''')












```
