```python
#Doubly linked List:
#operations:Insert left, insert right, delete left, delete right, insert after, delete specific, print
class Linked_list:
    def __init__(self):
        self.root=None
    def insert_left(self,n):
        if self.root==None:
            self.root=n;
        else:
            n.right=self.root#1
            self.root.left=n#2
            self.root=n;#3
    def insert_right(self,n):
        if self.root==None:
            self.root=n;
        else:
            t=self.root#1
            while t.right!=None:#2
                t=t.right
            t.right=n#3
            n.left=t#4

    def delete_left(self):
        if self.root==None:
            print("\nEmpty list")
        else:
            n=self.root#1
            self.root=self.root.right#2
            self.root.left=None#3
    def delete_right(self):
        if self.root==None:
            print("\nEmpty list")
        else:
            t=self.root#1
            while t.right!=None:#2
                t=t.right
            if t==self.root:#single node
                self.root=None
            else:
                t2=t.left#3
                t2.right=None
            print(f"{t.data} deleted")
    def print_elements(self):
        if self.root==None:
            print("\nEmpty list")
        else:
            n=self.root
            while n!=None:
                print(f"|{n.data}|",end='->')
                n=n.right
    def reverse_print_elements(self):
        if self.root==None:
            print("\nEmpty list")
        else:
            n=self.root
            while n.right!=None:
                n=n.right
            #n is on right most
            while n!=None:
                print(f"|{n.data}|",end='->')
                n=n.left
```

```python
class Node:
	def __init__(self,data):
		self.data=data
		self.left=self.right=None
```

```python
class Tree:
    #Object of tree is created, root should be created as a member.
    def __init__(self):
        self.root=None
    #Should return reference of a root from the class to the script.
    def get_root(self):
        return self.root

    def insert(self,r,n):#Reference to the root and a new node.
        print(n.data)
        if self.root==None:
            self.root=n#The first node becomes the root.
        else:
            if n.data<r.data:
                if r.left==None:
                    r.left=n
                    print("Inserted")
                else:
                    self.insert(r.left,n)
            else:#right side
                if r.right==None:
                    r.right=n
                    print("Inserted")
                else:
                    self.insert(r.right,n)
    def inorder(self,r):
        if r!=None:
            self.inorder(r.left)#L
            print(r.data,end="-")#P
            self.inorder(r.right)#R

    def count_nodes(self,r):
        if r==None:
            return 0
        return 1+self.count_nodes(r.left)+self.count_nodes(r.right)

    def count_leaf_nodes(self,r):
        if r==None:
            return 0
        if r.left==None and r.right==None:
          return 1
        return self.count_leaf_nodes(r.left)+self.count_leaf_nodes(r.right)

    def sum_nodes(self,r):
      if r==None:
        return 0
      return r.data+self.sum_nodes(r.left)+self.sum_nodes(r.right)

    def searching_key(self,r,key):#true if found else false
        if self.root==None:
            print("Empty Tree")
            return False
        else:
            if r==None:
                return False
            else:
                if key==r.data:
                    return True
                elif key<r.data:
                    return self.searching_key(r.left,key)

                else:
                    return self.searching_key(r.right,key)

```

```python
#code to implement Adjacency Matrix.
def print_graph(g):
    for row in g:
        print(row)

def create_graph(number_of_nodes):
    g=[0]
    g=g*number_of_nodes #rows
    for i in range(number_of_nodes):
        g[i]=[0]*number_of_nodes#columns
    for row in range(0,len(g)):
        for column in range(0,len(g)):
            print(f"from V{row} to V{column}:")
            g[row][column]=int(input(":"))
    return g
g=create_graph(int(input("Enter size of graph:")))
print_graph(g)
```
