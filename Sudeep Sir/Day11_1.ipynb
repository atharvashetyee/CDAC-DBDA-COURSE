```python
#Chi-square test for goodness of fit. 
"""
When there is a single categorical variable, this test is used. Obviously, with multiple categories. Categorical variable ek ho ka, but it can have 
multiple levels.Like, for example, "model of the car" is one variable. Levels could be Maruti, Hyundai, Kata, Mercedes, Volkswagen. So, this test is
used when there
is one categorical variable. The counts obtained from a sample (the counts for different levels or categories) need to be compared against the claimed
pattern. And establish whether the difference is statistically significant or not. 
The null hypothesis for these state is: the count follows the pattern. 
"""
```

```python
"""
Model of Car  pattern   counts
Maruti        0.35      120
TATA          0.3       105
Hyundai       0.15      80
Toyota        0.12      50
Mercedes      0.08      30
"""
```

```python
pattern=np.array([0.35,0.3,0.15,0.12,0.08])
obs_counts=np.array([120,105,80,50,30])
```

```python
sum(obs_counts) 
```

```python
exp_counts=pattern*sum(obs_counts) 
exp_counts
#Formula for Expected Count 
```

```python
scipy.stats.chisquare(obs_counts,f_exp=exp_counts)
```

```python
ts_array=(obs_counts-exp_counts)**2/(exp_counts)
ts_array
```

```python
np.sum(ts_array) #11.474953617810764
#This value is same as the test statistic. 
```

```python
#Category which made the most significant contribution is the one with the 8.572 value. In this case, that is Hyundai.
"""
Means that expected count for Hyundai is 50, but the observed count is 80. So, there is a significant difference.
 And we can observe one more thing in ts_array, the value in the last index is 0.020 which is smallest. It refers that there is very small difference between the observed and expected count of Mercedes. 
 You can see that the observed count is 30 and the expected count is 30.8. 
"""
```

# Chi-Square Test for Degree of Association. 


```python
"""Degree of association is used when you have two categorical variables. Number of variables for each. 
The number of vaiables for each categorical variable may not be the same. 
This test is used to check whether the choice made from the levels of one variable depends on the level of the second variable. ""'
```

```python
"""Null hypothesis The two variables Are not associated with each other.
        veg     Nonveg    Fastfood
male    110     150       40
female  120     80        90

"""
```

```python
data=np.array([[110,150,40],[120,80,90]])
```

```python
chi2_contingency(data)
"""
(Number of rows - 1) * (number of columns - 1) = degree of freedom. 

 expected_freq=array([[116.94915254, 116.94915254,  66.10169492],
       [113.05084746, 113.05084746,  63.89830508]]))
 if the choice of food was not associated with the gender, these are supposed to be the expected values. As we reject the null hypothesis,
 the food choice is dependent upon the gender. 

"""
```

```python
"""
The expected frequency is counted in the following way: manual calculation. 
for specific values like fastfood expected count of females
female(fast food)=total fast food* total female/total count

"""
```

```python
exp_values=chi2_contingency(data)[3]
ts=(exp_values-data)**2/exp_values
ts
```

```python
sum(sum(ts))
```
