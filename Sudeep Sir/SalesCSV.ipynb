Create a new column with the name Code. The value in Code column should be A if the TotalPrice is more than 60000. If it is between 10000 and 60000 then the value in Code column should be B. The value should be C if the value in TotalPrice is less then 10000


```python
Code=[]
list1=df['TotalPrice'].tolist()
for i in list1:
    if (i>60000):
        Code.append("A")
    elif(60000>i>10000):
        Code.append("B")
    else:
        Code.append("C")
```

```python
df['Code']=Code
```
