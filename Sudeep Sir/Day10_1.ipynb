# One-sample variance test 




"""
here it becomes that the variance of a sample has to be compared against the claim value.
and To establish whether their ratio is statistically significant or not. 
For continuous data 


"""




```python
df=pd.read_excel("CDAC_DataBook.xlsx",sheet_name='faithful')
```

```python
erupt=df.eruptions
```

```python
"""
Minimum value of variance is 1.42. 
H0:var(erupt)>=1.42
Ha:var(erupt)<1.42==> left tail test
"""
```

```python
ts=(len(erupt)-1)*np.var(erupt,ddof=1)/1.42  #Variance we have ==>np.var(erupt,ddof=1) expected variance==>1.42
```

```python
ts
```

```python
p_value=chi2.cdf(ts,len(erupt)-1) #We use CDF as its left tail test. 
p_value
```

```python
np.var(erupt,ddof=1)
```

```python
"""
Even the value of variation is less than 1.42 . We need to consider about the chance variation. 
The value of p is greater than LoS. We do not reject the null hypothesis. 
"""
```

```python
"""
Maximum value of variance is 1.1. 
H0:var(erupt)<=1.1
Ha:var(erupt)>1.1==> right tail test
"""
```

```python
ts=(len(erupt)-1)*np.var(erupt,ddof=1)/1.1
```

```python
p_value=chi2.sf(ts,len(erupt)-1)  #We use SF as its right tail test. 
p_value
```

```python
"""
In this scenario, we reject the null hypothesis as a value of p is less than LoS. 
So we can conclude that the value of variance is not less than 1.1. 
"""
```

```python
"""Hypothesis is that the value of erupt is equal to 1.4 or let's make it equal to 1.5, let's see what happens. So now my new hypothesis is 
that the value, this will become a true data, this is equal to 1.5, so this would be not equal to 1.5 and this would be 1.5 and this would be 1.5. 
This is what is happening."""

ts=(len(erupt)-1)*np.var(erupt,ddof=1)/1.5
```

```python
p1=chi2.cdf(ts,len(erupt)-1)
p1
```

```python
p2=chi2.sf(ts,len(erupt)-1)
p2
```

```python
p_value=p1*2 #We select the p_value which is smaller and multiply that by 2 to get the p_value. We do this because p_value can't be greater than 1 as
#probability can't be greater than 1
p_value
```
