```python
pune=pd.read_excel('CDAC_DataBook.xlsx',sheet_name='Pune')
mumbai=pd.read_excel('CDAC_DataBook.xlsx',sheet_name='Mumbai')
```

```python
pune
```

```python
mumbai
```

```python
pd.merge(pune,mumbai,on='Subject') #on is used to specified column that we want to take as common
#default value for "on" is all the subjects,for "how" that default value is inner
```

```python
pd.merge(pune,mumbai,on=['Subject','Grade']) #now it will take rows which are common in these both colmns
```

```python
pd.merge(pune,mumbai,on=['Subject'],how="outer") #when we give how='outer' it takes common value, those are present in both
```

```python
pd.merge(pune,mumbai,on=['Subject'],how="left") #gives priority to pune take all rows from pune and give nan if that is not present in mumbai
```

```python
df=pd.read_excel("CDAC_DataBook.xlsx",sheet_name="Health")
```

```python
df.head()
```

```python
df2=df.melt(id_vars=['Country','Disease'],var_name="Year",value_name="Cases")#id_var the colmn which i dont want to change,var_name=the name of new column,
#the column not present in id_vars is taken for adding in new coln, and values of that coln is stored in value_name 
```

```python
df2
```

```python
#creating more columns,from the df  we use given commands

df3=df2.pivot_table(columns='Disease',values='Cases',index=['Country','Year']).reset_index().rename_axis(columns=None)
df3
#it creates new colns from disease and by taking values from cases and written in index will be unchanged, then reset index gives index from
#0 and then rename axis is used to --->
```
