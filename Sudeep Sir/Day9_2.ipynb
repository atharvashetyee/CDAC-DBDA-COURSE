```python
#This is PAIRED T Test.

"""
This test is used To compare the average of paired differences between two samples (mostly before and after) with a claim value and establish
whether the difference is statistically significant or not. 

"""
```

```python
before=[5,8,7,6,9,8,7]
after=[4,9,5,5,9,8,6]
```

```python
#I claim that after the process of re-map, the individual TAT has come down by at least 0.7. min
```

```python
#Ho avg(before-after)>=0.7 mins
#Ha avg(before-after)<0.7 mins
```

```python
before=np.array([5,8,7,6,9,8,7])
after=np.array([4,9,5,5,9,8,6])
```

```python
ssw.ttost_paired(before,after,low=0.7,upp=0.7)
"""
-0.34856850115866744 -test statistics
0.6303407448408251- right tailed test answer
 0.3696592551591749 - left tailed test answer
"""
```

```python
scipy.stats.ttest_1samp(before-after,popmean=0.7,alternative='less')
```
