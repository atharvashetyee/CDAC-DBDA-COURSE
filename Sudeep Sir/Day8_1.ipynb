```python
'''
Steps involved in hypothesis testing

Step-1: Establish null (Ho) and alternative (H1 or Ha) hypothesis
Null hypothesis is made in such a way that if it is foind to be true
a. no action required
b. no decision required
c. no changes required

Case-1---salary example
Ho sal >= 10LPA
Ha sal < 10LPA
Error lies on the left or lower side
LEFT TAIL TEST

Case-2--- delivery example
Ho t <= 30
Ha t > 30
Error lies on the right or upper side
RIGHT TAIL TEST

Case-3--- salt in the food
Ho s = 2
Ha s not= 2
Error lies on the both sides
TWO TAIL TEST

Logical statement

Ho : I will not fall sick tomorrow
Ha : I will fall sick tomorrow

Ho: I will get a job from CDAC
Ha: I will not get a job from CDAC

Step-2 : Establish confidence level and Level of significance
Level of significane = 100 - confidence level

Step-3: Choose appropriate statistical test to perform the analysis

Step-4: Use the python functions or scripts to calculate test statistic
and p-value

Step-5: Compare the p-value (from step#4) against the LoS (from step#2)
IF THE P-VALUE IS LESS THAN LOS THEN WE REJECT THE HO
IF THE P-VALUE IS MORE THAN LOS THEN WE DO NOT REJECT THE HO

Type-I error
When we reject a null hypothesis which is actually true

Type-II error
When we do not reject a null hypothesis which is actually false

LoS is the maxiumum allowed probability of committing the Type-I error
p-value is the actual probability of committing the Type-I error

Ho : stay at home
Ha : move out

p-value : prob os catching infection is I move out (Arogya setu)
Los: max allowed prob of catching infection (family doctor) 0.05

pvalue>loc we do not reject the null hypothesis
pvalue<loc we reject the null hypothesis
pvalue=0 we reject null hyothesis

pvalue>0.05

Case-1
p-value = 0.2
This value is more than LoS. There is more chance of getting infection than
what the doctor prescribed. So we stay at home.

There is a 20% chance of rejecting a Ho which is actually true. 
Since the max allowed prob of rejecting something true is 5%
So we DO NOT take a risk and we DO NOT REJECT Ho


Case-2
p-value = 0.02
This value is less than LoS. There is less chance of getting infection than
what the doctor prescribed. So we reject staying at home and move out.

There is a 2% chance of rejecting a Ho which is actually true. 
Since the max allowed prob of rejecting something true is 5%
So we take a risk and we REJECT Ho


error on left==>salary
error on right==> any kind of delivery
two tailed==> salt in food,BMI

1-CL=L.O.S

reason behind taking wrong decision=as we are working on sample data
'''

```
