```python
df=pd.read_excel("CDAC_DataBook.xlsx",sheet_name='EmpInfo')
df
```

```python
ind=np.where(df.Deptt.isnull())
ind
ind2=np.where(df.EmpID.isnull())
ind2
```

```python
#one df--> values which are missing , and one df with clean df
ind3=np.where(df.EmpID.isnull() | df.Deptt.isnull()) #either dept or emp id is null
```

```python
newdf=df.dropna() #drop na values
newdf 

df.isnull() #boolean values of whether na or not
df.isnull().sum() #gives the sum of whether 
```

```python
df.Deptt.fillna('Overhead') #fills those to the coln NA ,by writing you mention in '' 
df['NewCol']=df.Deptt.fillna('Overhead') "add new coln and writes in place of null values"

#df.Deptt.fillna("Overhead",inplace=True) #to override the data coln without adding new
```

```python
x1=np.array([2,4,6,3,np.nan,3,1])
df['Marks']=x1
#we carry out median on the null values to fill the column when coln has intergers
#we carry out mode for column where those are null where it is string
```

```python
df['Marks']=df.Marks.fillna(df.Marks.median())
```

```python
df.Deptt.fillna(method="ffill") #the value before nan is filled at the place of nan
df.Deptt.fillna(method="bfill") # the value after nan is filled at the place of nan
```
